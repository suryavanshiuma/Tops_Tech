5. You received a dataset of ratings as strings from Spotify: ['4.5', '3.0', '5', '4.2']. Use type casting to convert these to floats, then find and print the highest rating.<br><br><em><strong>Hint:</strong> Use the float() function inside a loop or list comprehension.</em>

```python
rating = ['4.5', '3.0', '5', '4.2']

rating2 = []

for i in rating:
    j = float(i)
    rating2.append(j)

print(rating2)
highest_rating = max(rating2)
print("highest_rating : ", highest_rating)
```

2. Write a function clean_brand_name(name) that removes leading/trailing spaces and replaces any hyphens '-' with a single space in the input string. Test it with ' oneplus-Nord '.

```python
def clean_brand_name(name):
    cleaned = name.strip(name)          # .strip() → removes spaces from the beginning and end.
    cleaned = name.replace('-', ' ')

    return cleaned
```

5. Suppose you have a list of messy product names: [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']. Write code to clean each name (remove spaces, replace hyphens with spaces, and make the brand title case) and print the cleaned list.<br><br><em><strong>Constraint:</strong> Use at least three string methods from this session.</em>

```python
product_names = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']

cleaned_product_names = []

for name in product_names:
    clean_name = name.strip()
    clean_name = clean_name.replace("-", " ")
    clean_name = clean_name.title()

    cleaned_product_names.append(clean_name)

print(cleaned_product_names)
```
